## Task 2: The columns



### 2.1 Data dictionary

```python



dictionary = pd.DataFrame({
    'type':        df.dtypes.astype(str),
    'missing (%)': (df.isna().mean() * 100).round(1),
    'distinct':    df.nunique(),
})

meaning = {
    'n_tokens_content':             'Number of words in the article body.',
    'n_unique_tokens':              'Rate of unique words in the article body.',
    'n_non_stop_words':             'Rate of non-stop words in the article body.',
    'n_non_stop_unique_tokens':     'Rate of unique non-stop words in the article body.',
    'num_hrefs':                    'Number of links in the article.',
    'num_self_hrefs':               'Number of links to other articles published by Mashable.',
    'num_imgs':                     'Number of images in the article.',
    'num_videos':                   'Number of videos in the article.',
    'average_token_length':         'Average word length in the article body, in characters per word.',
    'num_keywords':                 'Number of keywords in the article metadata.',
    'global_subjectivity':          'Overall subjectivity score of the article body.',
    'global_sentiment_polarity':    'Overall sentiment polarity score of the article body.',
    'global_rate_positive_words':  'Proportion of all words in the article body that are positive.',
    'global_rate_negative_words':  'Proportion of all words in the article body that are negative.',
    'rate_positive_words':         'Proportion of non-neutral words that are positive.',
    'rate_negative_words':         'Proportion of non-neutral words that are negative.',
    'article_title':               'Article headline as free text.',
    'channel':                     'The news category assigned to the article.',
}


decision = {
    'n_tokens_content':             'keep',
    'n_unique_tokens':              'keep',
    'n_non_stop_words':             'keep',
    'n_non_stop_unique_tokens':      'keep',
    'num_hrefs':                    'keep',
    'num_self_hrefs':               'keep',
    'num_imgs':                     'keep',
    'num_videos':                   'keep',
    'average_token_length':         'keep',
    'num_keywords':                 'keep',
    'global_subjectivity':          'keep',
    'global_sentiment_polarity':    'keep',
    'global_rate_positive_words':   'keep',
    'global_rate_negative_words':   'keep',
    'rate_positive_words':          'keep',
    'rate_negative_words':          'keep',
    'article_title':                'drop: raw text outside the tabular feature scope',
    'channel':                     'target',
}


dictionary['meaning'] = dictionary.index.map(meaning).fillna('Meaning unclear')
dictionary['decision'] = dictionary.index.map(decision).fillna('Review needed')
with pd.option_context('display.max_colwidth', None):
    display(dictionary)
```
